```python
# 1- Valores nulos

nulos = df_limpo.isnull().sum()

print('Valores nulos por coluna: ')
print(nulos)
```

```python
# 1.1- Percentual de nulos

percentual_nulos = (df_limpo.isnull().sum()/len(df_limpo) * 100)

resultado_nulos = pd.DataFrame({
    "Quantidade de nulos": nulos,
    "Percentual (%)": percentual_nulos
})

print(resultado_nulos)
```

```python
# 2- Linhas Duplicadas

duplicados = df_limpo.duplicated().sum()

print(f'Quantidade de linhas duplicadas: {duplicados}')
```

```python
# 2.2- Visualizando as linhas duplicadas

df_limpo[df_limpo.duplicated()]
```

```python
# 3- Verificar os identificadores

# TRANSACTION ID

print(f'Total de registros: {len(df_limpo)}')
print(f'Transactions IDs únicos: {df_limpo['Transaction ID'].nunique()}')
print(f'Transaction IDs duplicado: {df_limpo['Transaction ID'].duplicated().sum()}')

# CUSTOMER ID

print(f'\nClientes únicos: {df_limpo['Customer ID'].nunique()}')
print(f'Customer IDs duplicado: {df_limpo['Customer ID'].duplicated().sum()}')
```

```python
# 3.2- Verificando dados repetidos

clientes_repetidos = df_limpo['Customer ID'].value_counts().loc[lambda serie: serie > 1]

clientes_repetidos
```

```python
# 4- Tipos de dados

df_limpo.dtypes
```

```python
# 5- Convertendo a coluna de data

df_limpo['Date'] = pd.to_datetime(df['Date'], errors='coerce')
```

```python
df_limpo["Date"].dtype # Verificando se foi convertido
```

```python
df_limpo["Date"].isnull().sum() # Verificando se as datas foram convertidas
```

```python
# 6- Padronizar colunas de texto

colunas_texto = ['Customer ID', 'Gender', 'Product Category']

for coluna in colunas_texto:
    df_limpo[coluna] = df_limpo[coluna].astype('string').str.strip()
```

```python
# 7- Verificando categorias de genero

df_limpo['Gender'].unique()
```

```python
df_limpo['Gender'].value_counts()
```

```python
# 8- Verificando categorias de produto

df_limpo['Product Category'].unique()
```

```python
df_limpo['Product Category'].value_counts()
```

```python
# 9- Garantir que as colunas numéricas estejam no tipo correto

colunas_numericas = [
    'Transaction ID',
    'Age',
    'Quantity',
    'Price per Unit',
    'Total Amount'
]

for coluna in colunas_numericas:
    df_limpo[coluna] = pd.to_numeric(df_limpo[coluna], errors='coerce')
    
```

```python
df_limpo.dtypes
```

```python
# 10- Verificando estatisticas das colunas numéricas

df_limpo[colunas_numericas].describe()
```

```python
# 11- Verificar valores suspeitos

print('Idade mínima: ', df_limpo['Age'].min())
print('Idade máxima: ', df_limpo['Age'].max())

df_limpo[
    (df_limpo['Age'] < 18) |
    (df_limpo['Age'] > 100)
]
```

```python
print("Quantidade mínima:", df_limpo["Quantity"].min())
print("Quantidade máxima:", df_limpo["Quantity"].max())

df_limpo[df_limpo["Quantity"] <= 0]
```

```python
print("Preço mínimo:", df_limpo["Price per Unit"].min())
print("Preço máximo:", df_limpo["Price per Unit"].max())

df_limpo[df_limpo["Price per Unit"] < 0]
```

```python
print("Valor mínimo:", df_limpo["Total Amount"].min())
print("Valor máximo:", df_limpo["Total Amount"].max())

df_limpo[df_limpo["Total Amount"] < 0]
```

```python
# 12- Verificando periodo das vendas

print("Data mínima:", df_limpo["Date"].min())
print("Data máxima:", df_limpo["Date"].max())
```

```python
# Transações por ano

df_limpo['Date'].dt.year.value_counts()
```

```python
# Transações por mês

df_limpo['Date'].dt.month.value_counts().sort_index()
```

```python
# 13- Criando coluna de ano e mês

df_limpo['Year'] = df_limpo['Date'].dt.year
df_limpo['Month'] = df_limpo['Date'].dt.month
df_limpo['Month Name'] = df_limpo['Date'].dt.month_name()
```

```python
df_limpo[
    ["Date", "Year", "Month", "Month Name"]
].head()
```

```python
# 14- Verficando se o valor total esta correto

df_limpo['Calculated Amount'] = (df_limpo['Quantity'] * df_limpo['Price per Unit'])
```

```python
df_limpo['Amount Difference'] = df_limpo['Total Amount'] - df_limpo['Calculated Amount']
```

```python
df_limpo['Amount Difference'].value_counts()
```

```python
# 15- Criando faixas etarias

df_limpo['Age Group'] = pd.cut(
    df_limpo['Age'],    
    bins = [17, 24, 34, 44, 54, 64],
    labels = ['18-24', '25-34', '35-44', '45-54', '55-64']
)
```

```python
df_limpo['Age Group'].value_counts().sort_index()
```

```python
# 16- Verificando novamente a qualidade após tratamento dos dados

print("Valores nulos após o tratamento:")
print(df_limpo.isnull().sum())
```

```python
print(
    "Duplicidades após o tratamento:",
    df_limpo.duplicated().sum()
)
```
